```python
import sys
import pandas as pd
import numpy as np, joblib
sys.path.append("/content/sample_data/")
from optimization import load_data, simulate_product
df=load_data("/content/Nassau_Candy_Cleaned.csv")
model=joblib.load("/content/best_model.pkl")
```

**Scenario engine**

The model is a pipeline, so raw categorical values are accepted directly. Region centroids are used only when customer coordinates are unavailable; this assumption must be stated in the report.

```python
# Choose a representative product and simulate all five factories.
product=df["Product Name"].dropna().iloc[0]
# The optimization helper expects an encoder argument for backward compatibility; the pipeline itself handles categories.
# For notebook experimentation, use the dashboard implementation in src/optimization.py as the canonical engine.
print(product)
```

```python
# Simple model-based what-if table for all factories using a representative order.
row=df[df["Product Name"]==product].iloc[0]
rows=[]
for fac in ["Lot's O' Nuts","Wicked Choccy's","Sugar Shack","Secret Factory","The Other Factory"]:
    x=pd.DataFrame([{
      "Ship Mode":row["Ship Mode"],"Region":row["Region"],"Division":row["Division"],
      "Factory":fac,"Sales":row["Sales"],"Units":row["Units"],"Cost":row["Cost"],
      "Gross Profit":row["Gross Profit"],"Profit Margin":row["Profit Margin"],
      "Cost Percentage":row["Cost Percentage"],"Unit Price":row["Unit Price"]}])
    rows.append((fac,float(model.predict(x)[0])))
pd.DataFrame(rows,columns=["Candidate Factory","Predicted Lead Time"]).sort_values("Predicted Lead Time")
```
